```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

Leo las imágenes en escala de grises con OpenCV. 

```python
images_path = "imgs/"
images = ["img1_tp.png", "img2_tp.png"]

fig_imgs, axs_imgs = plt.subplots(nrows=1, ncols=len(images))
fig_imgs.set_size_inches(10, 5)

for ax, img_file in zip(axs_imgs, images):
    img = cv2.imread(images_path + img_file, cv2.IMREAD_GRAYSCALE)
    ax.imshow(img, cmap="gray")
    ax.set_title(img_file)
    ax.axis("off")
```

Obtengo sus histogramas

```python

fig_hists, axs_hists = plt.subplots(nrows=1, ncols=len(images))
fig_hists.set_size_inches(10, 5)

#Seteo el bins a 64
for ax, img_file in zip(axs_hists, images):
    img = cv2.imread(images_path + img_file, cv2.IMREAD_GRAYSCALE)
    ax.hist(img.ravel(), bins=64, range=[0, 256])
    ax.set_title(f"Histograma {img_file}")
```

Conclusión:
- Los histogramas de las imagenes resultaron idénticos, por lo que podemos intuir que ambas tonalidades son similares, tanto de la flor como del gradiente.
- Por otro lado, se utilizaron 64 bins en los histogramas para reducir el ruido y facilitar la comparación entre las distribuciones de intensidades de las imágenes.

- Por otro lado, si podria ser util usar histogramas para detección/clasificación, pero va a depender mucho el modelo y el problema que vamos a atacar. Un ejemplo de donde podria servir es quizás para clasificar imagenes (dia vs noche).
